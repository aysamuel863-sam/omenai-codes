```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
font1 = {"family":"serif", "color":"blue", "size":20}
font2 = {"family":"serif", "color":"green", "size":15}

x_points = np.array([10, 20, 15])
y_points = np.array([20, 10, 30])

plt.title("X and Y", fontdict = font1, )
plt.xlabel("X, yeah X", fontdict = font2 )
plt.ylabel("Y, wait, Y?", fontdict = font2)
plt.grid(c = 'r', ls = '--', lw = 0.5)

plt.plot(x_points, y_points, '*-.y')
plt.show()
```

```python
y_points = np.array([3, 8, 1, 10])
plt.plot(y_points, marker = 'o', ms = 20, mec = 'r', mfc = 'b', ls = ':', lw = 2.0, c = 'g')
plt.show()
```

```python
x_points = np.array([0, 1, 2, 3])
y_points = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 1)
plt.plot(x_points, y_points)
plt.title("First")

x_points = np.array([0, 1, 2, 3])
y_points = np.array([10, 20, 30, 40])

plt.subplot(1, 2, 2)
plt.plot(x_points, y_points)
plt.title("Second")

plt.suptitle("Both")
plt.show()
```

```python
x = np.array([4, 2, 3, 2, 6, 8, 9, 10, 3, 5, 7])
y = np.array([73, 26, 32, 73, 92, 95, 91, 75, 62, 81, 61])
plt.scatter(x, y, c = 'g', cmap = 'viridis', s = 35, alpha = 0.3)
plt.colorbar()
plt.show()
```

```python
x = ["Apple", "Banana", "Cashew", "Orange"]
y = [25, 15, 20, 37]
colors = ["Red", "Green", "Yellow", "Orange"]
plt.bar(x, y, width = 0.2, color = colors)
plt.title("Fruits")
plt.show()
```

```python
x = ["Apple", "Banana", "Cashew", "Orange"]
y = [25, 15, 20, 37]
colors = ["Red", "Green", "Yellow", "Orange"]
plt.barh(x, y, height = 0.2, color = colors)
plt.title("Fruits")
```
