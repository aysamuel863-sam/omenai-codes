```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Admin\.vscode\csv\pandas_practice.csv")
```

```python
print(df.head())
print(df.shape)
print(df.columns)
print(df.dtypes)
```

```python
size1 = {"size":50}
size2 = {"size":35}
plt.plot(df["Student_Name"], df["Score"], marker = 'o', ls = '-.')
plt.xlabel("Student Names", c = 'r', fontdict = size2)
plt.ylabel("Scores", c = 'r', fontdict = size2)
plt.title("Names vs Scores", c = 'b', fontdict = size1)
plt.grid()
plt.show()
```

```python
size1 = {"size":50}
size2 = {"size":35}
avg_class = df.groupby("Class")["Score"].mean()
classes = ["SS1A", "SS1B", "SS1C", "SS1D"]
colors = ["Red", "Green", "Yellow", "Orange"]
plt.ylabel("Score", c = 'b', fontdict = size2)
plt.xlabel("Class", c = 'b', fontdict = size2)
plt.title("Average score by class", c = 'c', fontdict = size1)
plt.grid()
plt.bar(classes, avg_class, width = 0.2, color = colors)
```

```python
size1 = {"size":35}
size2 = {"size":25}
study_hours = df["Study_Hours"]
score = df["Score"]
plt.xlabel("Study Hours", c = 'b', fontdict = size2)
plt.ylabel("Score", c = 'b', fontdict = size2)
plt.title("Study Hours vs Score", c = 'c', fontdict = size1)

plt.grid()
plt.scatter(study_hours, score, c = 'g')
# From the graph, I understand that the greater the study hours doesn't mean the greater
# the score
```

```python
size1 = {"size":35}
size2 = {"size":25}
study_hours = df["Attendance_Percent"]
score = df["Score"]
plt.xlabel("Attendance Percent", c = 'b', fontdict = size2)
plt.ylabel("Score", c = 'b', fontdict = size2)
plt.title("Attendance Percent vs Score", c = 'c', fontdict = size1)

plt.grid()
plt.scatter(study_hours, score, c = 'g')
```

```python
avg_sub = df.groupby("Subject")["Score"].mean()
subjects = ["Biology", "Chemistry", "Computer Science", "English", "Mathematics", "Physics"]
colors = ["Red", "Green", "Blue", "Orange", "Hotpink", "Magnenta"]
size1 = {"size":35}
size2 = {"size":25}
plt.ylabel("Score", c = 'b', fontdict = size2)
plt.xlabel("Subject", c = 'b', fontdict = size2)
plt.title("Average score by Subject", c = 'c', fontdict = size1)
plt.grid()
plt.bar(subjects, avg_sub, width = 0.2)
# The highest performing subject is Physics
```

```python
scores = df.Score
plt.title("Students scores")
plt.hist(scores)
```
